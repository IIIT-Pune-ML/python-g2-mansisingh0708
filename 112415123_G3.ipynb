```
# Now, use zip to combine elements from multiple lists
list_of_names = ["Alice", "Bob", "Charlie"]
list_of_ages = [25, 30, 35]
list_of_occupations = ["Engineer", "Doctor", "Artist"]

# Zip combines corresponding elements from two or more iterables into tuples.
# It returns a zip object, which can be converted to a list of tuples or iterated over.
zipped_data = list(zip(list_of_names, list_of_ages, list_of_occupations))
print(f"Zipped data: {zipped_data}")

print("\nIterating through zipped data:")
for name, age, occupation in zip(list_of_names, list_of_ages, list_of_occupations):
    print(f"Name: {name}, Age: {age}, Occupation: {occupation}")
```

Now use a zip iterator to do the same.

Now state the differences between a zip and enumerate iterator.

As explained in the cell below, `enumerate()` adds a counter to an iterable and returns it as an enumerate object. This is useful when you need both the index and the value of items in a sequence.

`zip()`, on the other hand, takes multiple iterables and aggregates elements from each of them. It pairs up elements based on their position, forming tuples. This is useful when you need to process multiple lists in parallel, combining related data across them.

Python`enumerate()` and `zip()` are built-in functions that can be used on sequences.

`enumerate()` function can be use to iterate over indices and items of a list.

`zip()` function make an iterator that aggregates elements from each of the iterables. In simple words it can be used to iterate over two or more lists in parallel.

```
import statistics

# Open file
with open("/poem.txt", "r") as file:
    lines = file.readlines()

# 1. Length of each line
lengths = [len(line.strip()) for line in lines]

print("Length of each line:")
print(lengths)

# 2. Statistics
print("Sum:", sum(lengths))
print("Mean:", statistics.mean(lengths))
print("Median:", statistics.median(lengths))
print("Standard deviation:", statistics.stdev(lengths))

# Read entire file
with open("/poem.txt", "r") as file:
    text = file.read()

# 3. Number of words
words = text.split()

print("Number of words:", len(words))

# 4. First 1000 characters
print("First 1000 characters:")
print(text[:1000])

# 5. 3rd to 5th words
print("3rd to 5th words:")
print(words[2:5])

# 6. Every third word between 2nd and 15th
print("Every third word:")
print(words[1:15:3])

# 7. Last word
last_word = words[-1]

print("Last word:", last_word)

# Change second-last character
new_word = last_word[:-2] + "g" + last_word[-1:]

print("Modified word:", new_word)

# 8. Explanation
print("Direct modification was not successful.")
print("Reason: Strings are immutable in Python.")
```

# Working with CSV

Let's import a file with fuel economy data

```
import csv

mpg = []

with open('/mpg.csv', 'r') as csvfile:
    reader = csv.DictReader(csvfile)

    for row in reader:
        mpg.append(row)

print(mpg[:2])

```

Find the average cty fuel economy across all cars. All values in the dictionaries are strings.

```
total = 0

for d in mpg:
    total += float(d['cty'])

average_cty = total / len(mpg)

print("Average city MPG:", average_cty)
```

Find the average hwy fuel economy across all cars

```
total = 0

for d in mpg:
    total += float(d['hwy'])

average_hwy = total / len(mpg)

print("Average highway MPG:", average_hwy)
```

Group the cars by number of cylinder, and finding the average cty mpg for each group.

```
mpgByCyl = []

for c in cylinders:
    sumpg = 0
    count = 0

    for d in mpg:
        if d['cyl'] == c:
            sumpg += float(d['cty'])
            count += 1

    average = sumpg / count

    mpgByCyl.append((c, average))

print(mpgByCyl)
```

Return the unique values for the class types in our dataset using set.

```
vehicleclass = set(d['class'] for d in mpg)
```

Find average miles per gallon in the highway for each class of vehicles

```
MileageByClass = []

for t in vehicleclass:
    sumpg = 0
    count = 0

    for d in mpg:
        if d['class'] == t:
            sumpg += float(d['hwy'])
            count += 1

    average = sumpg / count
    MileageByClass.append((t, average))

MileageByClass.sort(key=lambda x: x[1])

print(MileageByClass)


```
